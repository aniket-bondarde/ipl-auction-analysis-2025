#Cleaning the Data (Matches Table)


```python
#Changing Date datatype
matches['date'] = pd.to_datetime(
    matches['date'],
    format='%d-%m-%Y'

)

#Replacing Null Value Of method
matches['method'] = matches['method'].fillna('Normal')


#Droping null values from Winner of Matches table
matches = matches.dropna(subset=['winner'])


#Replacing Null values in City
matches['city'] = matches['city'].fillna('Dubai')

#Replacing null values in result_margin
matches['result_margin'] = matches['result_margin'].fillna(0)



```

Cleaning the data for Deliveries Table

```python
#Replacing Null values
deliveries['extras_type'] = deliveries['extras_type'].fillna('no_extra')
deliveries['player_dismissed'] = deliveries['player_dismissed'].fillna('not_out')
deliveries['dismissal_kind'] = deliveries['dismissal_kind'].fillna('no_dismissal')

needs_fielder = ['caught', 'stumped', 'run out', 'caught and bowled']
no_fielder = ['bowled', 'lbw', 'hit wicket', 'no_dismissal']

deliveries['fielder'] = np.where(
    deliveries['dismissal_kind'].isin(no_fielder) & deliveries['fielder'].isna(),
    'no_fielder',
    np.where(
        deliveries['dismissal_kind'].isin(needs_fielder) & deliveries['fielder'].isna(),
        'unknown',
        deliveries['fielder']
    )
)

```

#Fixing Team Name


```python
team_name_change = {
    'Delhi Daredevils': 'Delhi Capitals',
    'Kings XI Punjab': 'Punjab Kings',
    'Royal Challengers Bangalore': 'Royal Challengers Bengaluru',
    'Rising Pune Supergiant': 'Rising Pune Supergiants'
    }
matches.replace(team_name_change, inplace=True)
deliveries.replace(team_name_change, inplace=True)

print("Team names updated")
```
