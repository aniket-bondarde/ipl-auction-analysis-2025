Merging The Data Sets

Cleaning season column

```python
merged['season'].apply(type).value_counts()
```

```python
merged.groupby('season').size()
```

Changing the datatype of column season

```python
#Changing everything into string.
merged['season'] = merged['season'].astype(str)




```

```python
#Extracting only years from the string 
merged['season'] = merged['season'].str.extract(r'(\d{4})')
```

```python
#Dropping null values
merged = merged.dropna(subset=['season'])
```

```python
#Changing everything into integer.
merged['season'] = merged['season'].astype(int)
```
